# Manuscript tables and figures

This notebook regenerates manuscript-facing summary tables and figure assets from the staged manuscript workflow.

```
import os
import sys
from pathlib import Path


def _find_repo_root() -> Path:
    env_root = os.environ.get("BSM_PUBLIC_RF_REPO_ROOT")
    if env_root:
        candidate = Path(env_root).resolve()
        if (candidate / "src").exists() and (candidate / "pyproject.toml").exists():
            return candidate

    for candidate in (Path.cwd(), *Path.cwd().parents):
        if (candidate / "src").exists() and (candidate / "pyproject.toml").exists():
            return candidate

    raise RuntimeError("Could not locate the bsm-public-rf repository root.")


REPO_ROOT = _find_repo_root()
SRC_DIR = REPO_ROOT / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))
```

```
from bsm_rfm import build_manuscript_notebook_context, run_final_manuscript_artifacts_stage

context = build_manuscript_notebook_context(REPO_ROOT, "08_manuscript_tables_and_figures.ipynb")
result = run_final_manuscript_artifacts_stage(context)
{
    "mode": context.runtime.mode,
    "model_performance": str(result.artifact_paths["model_performance"]),
    "workflow_stage_summary": str(result.artifact_paths["workflow_stage_summary"]),
    "figure_specs": str(result.artifact_paths["figure_specs"]),
}
```

### Step 1: Load Model Bundle and Generate Tables

Load model and coefficients from Stage 07 export bundle. Generate publication-ready summary tables.

**Tables generated:**
- Table 1: Model summary (R², sample count, feature count, AIC/BIC)
- Table 2: Top coefficients (ranked by |effect size|, with 95% CIs and p-values)
- Supplementary tables: all coefficients, feature importance, model diagnostics

**Export formats:**
- LaTeX (.tex): ready for manuscript
- CSV: for supplementary data
- HTML: for web version

**Parameters:**
- `table_format`: 'latex' (default), 'html', 'markdown'
- `top_n`: Number of top coefficients to highlight (default 20)

### Step 2: Generate Figures

Create publication-ready figures from model results and predictions.

**Figures generated:**
- Figure 1: Coefficient plot (forest plot, top effects with CIs)
- Figure 2: Predicted vs. observed scatter plot
- Figure 3: Residual diagnostics (Q-Q plot, scale-location plot)
- Supplementary figures: feature importance, stability analysis (if available)

**Export formats:**
- PNG (raster): 300 dpi, transparent background
- PDF (vector): publication-quality

**Parameters:**
- `figure_format`: 'png', 'pdf', or 'both' (default)
- `figure_size`: 'single-column' (3.5in), 'double-column' (7in), or custom
- `color_scheme`: 'publication' (grayscale/colorblind-safe, default), 'vibrant'
- `include_cis`: Include 95% confidence intervals (default True)
